```python
import numpy as np

# Student ID = 36
d1 = 6
d2 = 3

# Create Matrix A
A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
])

print("Matrix A:")
print(A)

# Shape
print("Shape:", A.shape)

# Determinant
det_A = np.linalg.det(A)
print("Determinant:", det_A)

# Rank
rank_A = np.linalg.matrix_rank(A)
print("Rank:", rank_A)

# Eigenvalues
eigen_A = np.linalg.eigvals(A)
print("Eigenvalues:", eigen_A)

# Inverse
if det_A != 0:
    inverse_A = np.linalg.inv(A)
    print("Inverse:")
    print(inverse_A)
```
